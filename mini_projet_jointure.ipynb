### Creation d'un DataFrame

```python
import pandas as pd
import numpy as np

# ============================
# SOURCE 1 : Ventes par magasin
# ============================
ventes_q1 = pd.DataFrame({
    'produit': ['Laptop', 'Smartphone', 'Tablette', 'Casque', 'Laptop', 
                'Smartphone', 'Tablette', 'Casque', 'Laptop', 'Smartphone'],
    'magasin': ['Paris', 'Paris', 'Paris', 'Paris', 'Lyon', 
                'Lyon', 'Lyon', 'Lyon', 'Marseille', 'Marseille'],
    'quantite': [45, 120, 30, 80, 35, 95, 25, 60, 50, 110],
    'chiffre_affaires': [45000, 60000, 12000, 4000, 35000, 
                         47500, 10000, 3000, 50000, 55000]
})

ventes_q2 = pd.DataFrame({
    'produit': ['Laptop', 'Smartphone', 'Tablette', 'Casque', 'Laptop',
                'Smartphone', 'Tablette', 'Casque', 'Tablette', 'Casque'],
    'magasin': ['Paris', 'Paris', 'Paris', 'Paris', 'Lyon',
                'Lyon', 'Lyon', 'Lyon', 'Marseille', 'Marseille'],
    'quantite': [50, 130, 35, 90, 40, 100, 28, 70, 30, 55],
    'chiffre_affaires': [50000, 65000, 14000, 4500, 40000,
                         50000, 11200, 3500, 12000, 2750]
})

# ============================
# SOURCE 2 : Infos magasins
# ============================
magasins_info = pd.DataFrame({
    'magasin': ['Paris', 'Lyon', 'Marseille', 'Bordeaux'],
    'region': ['Ile-de-France', 'Auvergne-Rhône-Alpes', 'PACA', 'Nouvelle-Aquitaine'],
    'surface_m2': [500, 350, 400, 300],
    'nb_employes': [25, 18, 20, 15]
})

# ============================
# SOURCE 3 : Infos produits
# ============================
produits_info = pd.DataFrame({
    'produit': ['Laptop', 'Smartphone', 'Tablette', 'Casque', 'Enceinte'],
    'categorie': ['Informatique', 'Téléphonie', 'Informatique', 'Audio', 'Audio'],
    'prix_unitaire': [1000, 500, 400, 50, 150],
    'fournisseur': ['TechCorp', 'MobilePlus', 'TechCorp', 'SoundMax', 'SoundMax']
})

# ============================
# SOURCE 4 : Objectifs par magasin/trimestre
# ============================
objectifs = pd.DataFrame({
    'magasin': ['Paris', 'Paris', 'Lyon', 'Lyon', 'Marseille', 'Marseille'],
    'trimestre': ['Q1', 'Q2', 'Q1', 'Q2', 'Q1', 'Q2'],
    'objectif_ca': [130000, 140000, 100000, 110000, 110000, 120000]
})

# ============================
# SOURCE 5 : Retours clients
# ============================
retours = pd.DataFrame({
    'produit': ['Laptop', 'Smartphone', 'Casque', 'Laptop', 'Tablette', 'Smartphone'],
    'magasin': ['Paris', 'Lyon', 'Paris', 'Marseille', 'Lyon', 'Paris'],
    'trimestre': ['Q1', 'Q1', 'Q2', 'Q2', 'Q1', 'Q2'],
    'nb_retours': [3, 5, 8, 2, 1, 7]
})

print("✅ Données créées avec succès !")
```

##### Exercice A1 : Concatèner les ventes Q1 et Q2

```python
ventes_all = pd.concat([ventes_q1, ventes_q2], keys=['Q1', 'Q2'])
print(ventes_all)
print(ventes_all.index)
```

```python
#On regroupe les doublons en faisant la somme Q1 + Q2

ventes_grouped = ventes_indexed.groupby(level=['magasin', 'produit']).sum()
ventes_grouped['quantite'].unstack('produit')
```
