#### Partie B - Fusion (merge)

- a) Fais un merge de ventes_all avec magasins_info sur la colonne 'magasin'
- b) Utilise how='left' pour garder toutes les ventes
- c) Stocke dans "ventes_enrichies"
- d) Question : Bordeaux apparaît-il ? Pourquoi ?

```python
# on remet les données comme avant 
ventes_all = pd.concat([ventes_q1, ventes_q2], ignore_index=True)
```

```python
ventes_enrichies = pd.merge(ventes_all, magasins_info, on='magasin', how='left')
print(ventes_enrichies)
```

```python
print(f"\nNombre de lignes : {len(ventes_enrichies)}")
print(f"\nBordeaux présent ? {'Bordeaux' in ventes_enrichies['magasin'].values}")
```

magasins_info contient Bordeaux mais ventes_all ne contint Aucune vente à Bordeaux
how='left' 
- on garde tout ce qui est à GAUCHE (ventes_all)
- Bordeaux n'est PAS dans les ventes
- donc Bordeaux N'APPARAÎT PAS dans le résultat

#### Difference clé entre les types de jointure :

- how='left'  → garde TOUTES les ventes, même sans info magasin
- how='right' → garde TOUS les magasins, même sans ventes (Bordeaux apparaîtrait)
- how='inner' → garde SEULEMENT les correspondances des deux côtés
- how='outer' → garde TOUT de partout


- EXERCICE B2 : Fusion ventes + infos produits
- Ajouter les infos produits à ventes_enrichies
- Utiliser how='left'
- Question : Le produit "Enceinte" apparaît-il ? Pourquoi ?

```python
# MERGE INNER
merge_inner = pd.merge(ventes_enrichies, produits_info, on='produit', how='inner')
print("INNER")
print(f"Lignes : {len(merge_inner)}")
print(f"Produits : {merge_inner['produit'].unique()}")
print()

# MERGE LEFT
merge_left = pd.merge(ventes_enrichies, produits_info, on='produit', how='left')
print("LEFT")
print(f"Lignes : {len(merge_left)}")
print(f"Produits : {merge_left['produit'].unique()}")
print()

# MERGE RIGHT
merge_right = pd.merge(ventes_enrichies, produits_info, on='produit', how='right')
print("RIGHT")
print(f"Lignes : {len(merge_right)}")
print(f"Produits : {merge_right['produit'].unique()}")
print()

# MERGE OUTER
merge_outer = pd.merge(ventes_enrichies, produits_info, on='produit', how='outer')
print("OUTER")
print(f"Lignes : {len(merge_outer)}")
print(f"Produits : {merge_outer['produit'].unique()}")
```

```python
print(merge_right[merge_right['produit'] == 'Enceinte'])
```

- On choisit INNER (ou LEFT, même reultat)

car on veut analyser des ventes réelles, Enceinte n'a aucune vente donc ça ne sert à rien de le garder

```python
# --- On met à jour ventes_enrichies pour la suite ---
ventes_enrichies = pd.merge(ventes_enrichies, produits_info, on='produit', how='inner')

print("\n=== Résultat final ===")
print(f"Shape : {ventes_enrichies.shape}")
print(f"Colonnes : {ventes_enrichies.columns.tolist()}")
print(ventes_enrichies.head())
```

- EXERCICE B3 : Exploration des types de join

```python
# a) INNER
merge_inner = pd.merge(ventes_all, magasins_info, on='magasin', how='inner')
print(f"INNER : {len(merge_inner)} lignes")
print(f"Magasins : {sorted(merge_inner['magasin'].unique())}")

# b) RIGHT
merge_right = pd.merge(ventes_all, magasins_info, on='magasin', how='right')
print(f"\nRIGHT : {len(merge_right)} lignes")
print(f"Magasins : {sorted(merge_right['magasin'].unique())}")
# Bordeaux apparaît avec des NaN car il n'a aucune vente
print("\n=== Ligne Bordeaux ===")
print(merge_right[merge_right['magasin'] == 'Bordeaux'])

# c) OUTER
merge_outer = pd.merge(ventes_all, magasins_info, on='magasin', how='outer')
print(f"\nOUTER : {len(merge_outer)} lignes")
print(f"Magasins : {sorted(merge_outer['magasin'].unique())}")
```

- INNER : 20 lignes → Bordeaux est exclu (pas de ventes)
- RIGHT : 21 lignes → Bordeaux apparaît avec NaN (quantite, CA = NaN)
- OUTER : 21 lignes → même résultat que right ici car tous les magasins de ventes_all existent dans magasins_info
- Conclusion : pour analyser des ventes réelles, on utilise INNER ou LEFT
- car garder Bordeaux sans données de vente ne sert à rien
